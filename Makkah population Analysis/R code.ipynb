Here we worked to arrange the data to make it esay to read for us

```python
library(readr) # library for function read Data

mydata <- read_csv("sample_data/makkah_data.csv") # Call Data
#print(mydata)

data_of_makkah <- subset(mydata, Region == "Makkah") # Value Allocation


library(dplyr) # library for function arrange Data

data_of_makkah <- data_of_makkah %>% arrange(desc(Year)) # arrange data

print(data_of_makkah, n = nrow(data_of_makkah)) # function for print all of rows
write.csv(data_of_makkah, file = "makkah_data.csv", row.names = FALSE) # save the Work
```

statastic for Data

```python
#statastic for problem 1

saudis <- aggregate(`Population estimates` ~ Year, 
                    data = subset(data_of_makkah, Nationality == "Saudi"), 
                    sum) # numbers for saudi from 2010 to 2022


non_saudis <- aggregate(`Population estimates` ~ Year, 
                        data = subset(data_of_makkah, Nationality == "Non-Saudi"), 
                        sum) # numbers for non-saudi from 2010 to 2022

summary(saudis$`Population estimates`)
sd(saudis$`Population estimates`)
var(saudis$`Population estimates`)

summary(non_saudis$`Population estimates`)
sd(non_saudis$`Population estimates`)
var(non_saudis$`Population estimates`)
```

```python
#statastic for problem 2

males <- aggregate(`Population estimates` ~ Year, 
                   data = subset(data_of_makkah, Gender == "Male"), 
                   sum)

summary(males$`Population estimates`)
sd(males$`Population estimates`)
var(males$`Population estimates`)

```

```python
# statastic for problem 3
females <- aggregate(`Population estimates` ~ Year, 
                     data = subset(data_of_makkah, Gender == "Female"), 
                     sum)

summary(females$`Population estimates`)
sd(females$`Population estimates`)
var(females$`Population estimates`)
```

```python
#statastic for problem 4
saudi_females <- aggregate(`Population estimates` ~ Year, 
                           data = subset(data_of_makkah, Gender == "Female" & Nationality == "Saudi"), 
                           sum)

summary(saudi_females$`Population estimates`)
var(saudi_females$`Population estimates`)
sd(saudi_females$`Population estimates`)

# غير السعوديات
non_saudi_females <- aggregate(`Population estimates` ~ Year, 
                               data = subset(data_of_makkah, Gender == "Female" & Nationality == "Non-Saudi"), 
                               sum)

summary(non_saudi_females$`Population estimates`)
sd(non_saudi_females$`Population estimates`)
var(non_saudi_females$`Population estimates`)
```

Visulasation

```python
# Load the tidyverse library for visualization and data manipulation
library(tidyverse)

# Create a data frame with the population data and years
library(tidyverse)

data <- data_of_makkah %>%
  group_by(Year, Nationality) %>%
  summarise(Population = sum(`Population estimates`), .groups = "drop")

ggplot(data, aes(x = Year, y = Population, fill = Nationality)) +
  geom_bar(stat = "identity", position = "dodge", width = 0.8) +
  scale_fill_manual(values = c("Saudi" = "gray", "Non-Saudi" = "red")) +
  labs(title = "Population Distribution by Year (Saudi vs Non-Saudi)",
       x = "Year",
       y = "Population") +
  theme_minimal()












```

```python

# Create a data frame with the population data and years
library(ggplot2)

males <- data_of_makkah %>%
  filter(Gender == "Male") %>%
  group_by(Year) %>%
  summarise(Population = sum(`Population estimates`), .groups = "drop")

ggplot(males, aes(x = Year, y = Population)) +
  geom_line(color = "blue", size = 1.2) +
  geom_point(color = "darkblue", size = 2) +
  labs(title = "Population of Males by Year",
       x = "Year",
       y = "Population") +
  theme_minimal()
```

```python

females <- data_of_makkah %>%
  filter(Gender == "Female") %>%
  group_by(Year) %>%
  summarise(Population = sum(`Population estimates`), .groups = "drop")

ggplot(females, aes(x = Year, y = Population)) +
  geom_line(color = "purple", size = 1.2) +
  geom_point(color = "darkred", size = 2) +
  labs(title = "Population of Females by Year",
       x = "Year",
       y = "Population") +
  theme_minimal()
```

```python

female_nationality <- data_of_makkah %>%
  filter(Gender == "Female") %>%
  group_by(Year, Nationality) %>%
  summarise(Population = sum(`Population estimates`), .groups = "drop")

ggplot(female_nationality, aes(x = Year, y = Population, color = Nationality)) +
  geom_line(size = 1.2) +
  geom_point(size = 2) +
  scale_color_manual(values = c("Saudi" = "blue", "Non-Saudi" = "red")) +
  labs(title = "Population of Saudi and Non-Saudi Women Over the Years",
       x = "Year",
       y = "Population") +
  theme_minimal()
```

```python

male_nationality <- data_of_makkah %>%
  filter(Gender == "Male") %>%
  group_by(Year, Nationality) %>%
  summarise(Population = sum(`Population estimates`), .groups = "drop")

ggplot(male_nationality, aes(x = Year, y = Population, color = Nationality)) +
  geom_line(size = 1.2) +
  geom_point(size = 2) +
  scale_color_manual(values = c("Saudi" = "blue", "Non-Saudi" = "red")) +
  labs(title = "Population of Saudi and Non-Saudi Men Over the Years",
       x = "Year",
       y = "Population") +
  theme_minimal()
```
