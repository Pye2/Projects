```python
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
import pandas as pd
```

```python
# read show_name_by_gender
show_name_by_gender = pd.read_csv(r"file path" ,encoding = "windows-1256")
dataframe = pd.DataFrame(show_name_by_gender)
dataframe
```

```python

# this code because the Messages of wornings
warnings.simplefilter(action='ignore', category=FutureWarning)



#count of gender and classifcation it
gender_counts = dataframe['SpeakerGender'].value_counts()

#plot by use seaporn 
sns.set(style="whitegrid")
plt.figure(figsize=(8, 6))
sns.countplot(x='SpeakerGender', data=dataframe, hue='SpeakerGender', palette='viridis', dodge=False)

# here use options for discribe visulization  
for i, count in enumerate(gender_counts):
    plt.text(i, count + 10, str(count), ha='center', va='bottom', fontsize=12, color='black')
# discribtion on the x axis & y axis 
plt.title('Distribution of Speaker Gender')
plt.xlabel('Speaker Gender')
plt.ylabel('Count')

plt.show()
```

```python
# Print the counts of gender
print("Counts for each gender :")
print(gender_counts)
```

```python

# note , call dataframe from show_name_by_gender 


# Count each gender for each show name
gender_counts_by_show = dataframe.groupby(['ShowName', 'SpeakerGender']).size().unstack(fill_value=0)

# plot use seaborn
sns.set(style="whitegrid")
plt.figure(figsize=(12, 12))
gender_counts_by_show.plot(kind='bar', stacked=True, colormap='viridis')
# discribtion on the x axis & y axis 

plt.title('Distribution of Speaker Gender by Show Name')
plt.xlabel('Show Name')
plt.ylabel('Count')
plt.legend(title='Speaker Gender')

plt.show()

```

```python
# read percent_of_procssedtexts file

percent_of_procssedtexts = pd.read_csv(r"file path" , encoding = 'windows-1256')

dataframe5 = pd.DataFrame(percent_of_procssedtexts)

dataframe5
```

```python







# Correcting the column name for 'ProcessedText' if necessary
if 'ProcessedText;' in dataframe5.columns:
    dataframe5 = dataframe5.rename(columns={'ProcessedText;': 'ProcessedText'})

# Remove potential trailing characters (like semicolons) in 'ProcessedText'
dataframe5['ProcessedText'] = dataframe5['ProcessedText'].str.rstrip(';')

# Calculate the number of processed texts
processed_count = (dataframe5['GroundTruthText'] != dataframe5['ProcessedText']).sum()

# Calculate the total number of texts
total_count = len(dataframe5)

# Calculate the percentage of processed texts
processed_percentage = (processed_count / total_count) * 100

# Output the result
print(f"Percentage of processed texts: {processed_percentage:.2f}%")
```

```python


# here we want to visulization the Percentage of processed texts vs unprocessed texts

# Calculate the percentage of unprocessed texts
unprocessed_percentage = 100 - processed_percentage

# Data to plot
labels = 'Processed Texts', 'Unprocessed Texts'
sizes = [processed_percentage, unprocessed_percentage]
colors = ['lightcoral', 'lightskyblue']
explode = (0.1, 0)  # explode 1st slice for emphasis

# Plotting the pie chart
plt.figure(figsize=(8, 8))
plt.pie(sizes, explode=explode, labels=labels, colors=colors,
        autopct='%1.1f%%', shadow=True, startangle=140)

plt.axis('equal')  # Equal aspect ratio ensures the pie chart is circular.
plt.title('Percentage of Processed vs Unprocessed Texts')
plt.show()
```
