```python
import requests
from bs4 import BeautifulSoup
import pandas as pd

# Define the function to scrape pages
def scrape_amazon_page(page_number):
    url = f"https://www.amazon.sa/-/en/s?i=beauty&rh=n%3A16630489031&fs=true&page={page_number}&language=en_AE&qid=1715857970&ref=sr_pg_{page_number}"
    response = requests.get(url)
    soup = BeautifulSoup(response.content, 'html.parser')
    
    product_data = []

    # Find all the product containers
    containers = soup.find_all("div", class_="puis-card-container s-card-container s-overflow-hidden aok-relative puis-expand-height puis-include-content-margin puis puis-v3nmsu0aru7pqu2fcvqir777xon s-latency-cf-section puis-card-border")
    
    for container in containers:
        product = {}

        # Extract product name
        name_span = container.find("span", class_="a-size-base-plus a-color-base a-text-normal")
        product['name'] = name_span.text if name_span else 'N/A'
        
        # Extract product price
        price_span = container.find("span", class_="a-price-whole")
        product['price'] = price_span.text if price_span else 'N/A'

        # Extract price after discount or indicate no discount
        discount_price_span = container.find("span", class_="a-price a-text-price")
        product['discount_price'] = discount_price_span.text if discount_price_span else 'No discount'

        # Extract product rating
        rating_span = container.find("span", class_="a-icon-alt")
        product['rating'] = rating_span.text if rating_span else 'N/A'

        # Extract number of reviews
        reviews_span = container.find("span", class_="a-size-base s-underline-text")
        product['reviews'] = reviews_span.text if reviews_span else 'N/A'
        
        # Extract options
        options_span = container.find("span", class_="a-truncate")
        product['options'] = options_span.text.strip() if options_span else 'N/A'

        product_data.append(product)
    
    return product_data

# Scrape pages 
all_product_data = []
for page in range(2, 400):
    all_product_data.extend(scrape_amazon_page(page))

# Convert the data into a pandas DataFrame
df = pd.DataFrame(all_product_data)

# Save the DataFrame to a CSV file
df.to_csv('amazon_products.csv', index=False, encoding='utf-8')

print(f"Data saved to amazon_products.csv")
```
